### Example 5 for the usage of Rela²x.
### See README.md for more information.

- ¹H and ¹³C nuclei.

- Zeeman interaction (including the chemical shift) and J-coupling.

- The Hamiltonian superoperator.

Import Rela²x.

```python
from rela2x import *
```

Define the spin system.

```python
spin_system = ['1H', '13C']
```

Define the coherent interactions.

```python
coh_intrs = {'Z': [1, 1],
             'J': [[0, 1],
                   [0, 0]]}
```

Compute the Hamiltonian commutation superoperator in the direct product basis of spherical tensor operators and create a HamiltonianSuperoperator object.

```python
H = hamiltonian_superoperator(spin_system, coh_intrs)
```

Display $H$ and visualize.

```python
display(H.op)
H.visualize(basis_symbols=H.basis_symbols)
```

Individual elements are read out with `frequency`, the coherent counterpart of `rate`. The diagonal elements are the coherence precession frequencies; $\hat T_{10}$ (longitudinal magnetisation) does not precess.

```python
display(H.frequency('110'))
display(H.frequency('111'))
```

Now repeat for a homonuclear pair. The J-coupling flip-flop term is secular for a homonuclear pair, so it stays in the Hamiltonian instead of being dropped.

```python
H_hom = hamiltonian_superoperator(['1H', '1H'], coh_intrs)
```

Display and visualize, exactly as above. The extra structure from the flip-flop term is visible directly in the nonzero pattern.

```python
display(H_hom.op)
H_hom.visualize(basis_symbols=H_hom.basis_symbols)
```
